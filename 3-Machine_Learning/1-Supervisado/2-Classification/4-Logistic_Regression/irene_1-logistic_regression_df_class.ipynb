## Creamos el modelo

```python
X = np.array(dataframe.drop(columns=['clase']))
y = np.array(dataframe['clase'])
```

```python
X = dataframe.drop("clase",axis=1).values
y = dataframe["clase"].values
```

```python
y
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X, y)
```

```python
predictions = model.predict(X)
print(predictions)
```

```python
predicions_proba = model.predict_proba(X)
print(np.round(np.array(predicions_proba), 2))
#predicions_proba


# prediciones
'''
predicicones de cada usuario, cada fila es un usuario, cada columna es la probabilidad de ser Windows, Linux o Mac respectivamente
[0.22 0.09 0.69], más prob mac
 [0.   0.   1.  ], mac, (sale 1 clavao porque aplicó un redondeo, np.round)
 [0.   0.   1.  ], mac
 [0.18 0.08 0.74], mac
 [0.28 0.1  0.62], mac
 [0.27 0.08 0.65], mac
 '''
```

```python
model.score(X, y)

# score es el f1, metrica combinada entre precision y recall
```

```python
model.classes_
```

# Validación del Modelo

```python
validation_size = 0.20
seed = 7
X_train, X_test, Y_train, Y_test = model_selection.train_test_split(X,
                                                                    y,
                                                                    test_size=validation_size,
                                                                    random_state=seed)
```

```python
name='Logistic Regression'
kfold = model_selection.KFold(n_splits=10) #Parte los datos en 10 trozos para usar validación cruzada / cross validation
cv_results = model_selection.cross_val_score(model, X_train, Y_train, cv=kfold, scoring='accuracy') # scoring es en lo que quiero que se fije el modelo, en este caso accuracy, pero podría ser precision, recall, f1, etc

msg = "%s: %f (%f)" % (name, cv_results.mean(), cv_results.std())
print(cv_results)
print(msg)
```

```python
# para cada una de esas 10 veces sale una accuracy de 78%, 64%, 71%....
# la media de los 10 sería un 72,8571% (el 0.094186 es la desviación estándar (media de las desviaciones), que nos dice cuánto varían los resultados entre las 10 veces))
# mi modelo acierta el 72 % de las veces, pero varía entre un 78% y un 66% dependiendo de qué datos le toque en cada una de las 10 veces (cada vez le toca un 20% de los datos para testearlo, y el otro 80% para entrenarlo)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)

# te sale cómo están por defecto los pa´rametros del modelo
# pero es en este caso que no los he tocado
```

```python
predictions = model.predict(X_test)
print(accuracy_score(Y_test, predictions))
predictions
```

```python
acierto = accuracy_score(Y_test, predictions)

error = 1 - acierto
print("Acierto:", round(acierto*100, 2), "%")
print("Error:", round(error*100, 2), "%")
```
