## Mini Hundir la Flota
#### Primeros pasos
Vamos a empezar a programar trozos de código que nos servirán para el HLF:

* Crear un Tablero como una lista de listas de 10x10 que contenga caracteres espacio: " "
* Insertar dos barcos sustituyendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haya una en "B")
        * Si hay una "B":
            * Cambiar el contenido de esa posición por una "x"
            * Imprimir por pantalla el texto "Tocado en posición `i,j`"
        * Si hay un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el texto "Agua"
    * Imprimir por pantalla el Tablero
* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una par i y otra par j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
    


```python
import numpy as np
```

```python
# crear tablero:
tablero = np.full((10, 10), " ", dtype=str) # creo la matriz 10x10 con espacios, añado el dtype para asegurarme que no habrá problema cuando meta la 'x' y la 'o'
print(tablero)
```

```python
# barquitos:

tablero[1, 0:4] = "B" # coloco elbarco horizontal en las columnas 0, 1, 2 y 3 de la fila 1

tablero[3:6, 3] = "B" # coloco el barco en las filas 3, 4 y 5 de la columna 3

print(tablero) # el resto del tablero sigue vacío
```

```python
# hago un while para que seguirá pidiendo coordenadas hasta que el usuario introduzca una válida
while True:
    ''' solicito las coordenadas al usuario:
    (como en python las posiciones de las listas empiezan en el 0, aviso al
    jugador de que las coordenadas empiezan en el 0 y acaban en 9 en
    este tablero de 10x10) '''
    i = int(input("Introduce la coordenada i (fila) del 0 al 9: "))
    j = int(input("Introduce la coordenada j (columna) del 0 al 9: "))

    # compruebo que las coordenadas están dentro del tablero:
    if not (0 <= i < 10 and 0 <= j < 10):
        print("Coordenadas fuera del tablero.")
        continue  # vuelvo a pedir coordenadas
    else: # si las coordenadas están dentro del tablero pasa lo siguiente:
        # comprobruebo las coordenadas del disparo:
        if tablero[i, j] == "B": # si en la coordenada hay una 'B', es decir, parte de la eslora de uno de los barcos, sucede lo siguiente:
            tablero[i, j] = "x" # se sustituirá la 'B' por una 'x' (marca del impacto)
            print(f"Tocado en posición {i},{j}.")
            print("Vuelves a disparar.")
            print(tablero)
            continue # como acertó, vuelve a disparar y continúa el while
        elif tablero[i, j] == " ": # si en la coordenada no hay nada, es decir, hay agua, sucede lo siguiente:
            tablero[i, j] = "o" # saldrá una 'o' para indicar que es agua (marca del fallo)
            print("Agua")
            print(tablero)
            break  # termina su turno porque falló
        else: # si no hay agua ni barco, es que ya ha dado esa coordenada anteriormente
            print("Esa casilla ya había sido disparada.")
            print(tablero)
            break # pierde turno por no haber estado atento a dónde disparaba
```

```python
# voy a intentar hacer el bonus de las funciones, pero no prometo que salga muy bien

# una función para solicitar coordenadas válidas:
def pedir_coordenada(mensaje):
    while True:
        try: # pedimos un número entero usando el mensaje recibido como argumento
            num = int(input())
            if 0 <= num <= 9: # compruebo si el número está dentro del rango permitido (0 a 9)
                return num  # si es válido, devolvemos la coordenada y salimos de la función
            else: # si el número está fuera del rango, aviso y se repite el bucle
                print("Coordenada fuera del rango 0-9.")
        except ValueError: # si el se introduce algo que no es un número entero:
            print("Debes introducir un número entero.") # sale un aviso
```

```python
# una función 'disparar' que devuelva 'True' si el usuario ha acertado o 'False' si no:
def disparar(tablero, i, j):
    if tablero[i, j] == "B":
        tablero[i, j] = "x"
        print(f"Tocado en posición {i},{j}.")
        print(tablero)
        return True # devolvemos True porque acertó
    elif tablero[i, j] == " ":
        tablero[i, j] = "o"
        print("Agua")
        print(tablero)
        return False # devolvemos False porque no acertó
    else:
        print("Esa casilla ya había sido disparada.")
        print(tablero)
        return False # fallo, no repite turno
```

```python
# turno del jugador usando las funciones:
while True:
    i = pedir_coordenada("Introduce la coordenada i (fila) del 0 al 9: ")
    j = pedir_coordenada("Introduce la coordenada j (columna) del 0 al 9: ")

    acierto = disparar(tablero, i, j) # llamo a la función disparar y guardo si acertó o no

    if acierto: # si acierta, repite turno
        print("Vuelves a disparar.")
        continue  # vuelve al inicio del while 
    else: # si falla o repite casilla, termina su turno
        break # sale del bucle
```
