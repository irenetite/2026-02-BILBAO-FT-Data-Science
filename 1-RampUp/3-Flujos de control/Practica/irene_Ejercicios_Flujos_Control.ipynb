## Ejercicio 1
Declara una variable numérica que será una hora cualquiera del día. Implementa la siguiente casuística:

* Si es entre las 0 y las 8, print "Durmiendo"
* Si es entre las 9 y las 18, print "Trabajando"
* Si es entre las 19 y las 21, print "Clase"
* Si es entre las 22 y las 24, print "Descanso"
* En cualquier otro caso, print "Transporte o error"

```python
hora = int(input("¿Qué hora es? Introdúce sólo la hora en formato 24 horas y no los minutos"))

if hora >= 0 and hora <= 8:
    print("Deberías estar durmiendo.")
    
elif hora >= 9 and hora <= 18:
    print("Ponte a trabajar.")
    
elif hora >= 19 and hora <= 21:
    print("Tira pa clase.")
    
elif hora >= 22 and hora <= 24:
    print("Descansa, te lo mereces.")
    
else:
    print("Error.")
```

En este ejercicio vamos a implementar un calculador de precios de casas muy sencillo. Tenemos las siguientes variables:

> superficie
>
> distrito

Implementa la siguiente casuística:

1. Si el distrito es "Moncloa" o "Centro", y además la superficie es superior a 100 metros cuadrados, el precio de la casa es de 1000
2. Si el distrito es "Salamanca", y además la superficie de la casa es al menos de 150 metros, el precio de la casa es de 1500
3. Si el distrito no es "Retiro" y la superficie está entre 60 y 80 metros, el precio es de 600
4. En cualquier otro caso, el precio será de 0

```python
superficie = int(input("Dime la superficie del inmueble:")) 
distrito = input("Ahora dime el distrito del inmueble:")

if (distrito == "Moncloa" or distrito == "Centro") and superficie > 100:
    print("El precio de la casa es de: 1000 euros.")
elif distrito == "Salamanca" and superficie >= 1500:
    print("El precio de la casa es de 1500 euros.")
elif distrito != "Retiro" and (superficie >= 60 or superficie <= 80):
    print("El precio es de 600 euros.")
else:
    print("No tenemos dato de eso.")
```

## Ejercicio 6
Dada la siguiente lista:

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que los recorra e imprima por pantalla todos los divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

for num_list1 in list1: # tengo que recorrer cada elemento uno a uno
    if num_list1 > 150: # si el número es que 150 se detiene el bucle
        break
    elif num_list1 % 5 == 0: # ahora compruebo si es divisible entre 5
        print(num_list1)
```

## Ejercicio 7
Escribe un programa en Python que imprima por pantalla todos los números que cumplan que sean divisibles por 5 y divisibles por 7, dentro del rango de valores (150, 350).

```python
rang_valores= range(150, 350) # defino el rango

for i in rang_valores: # tengo que recorrer los números dentro del rango
    if i % 5 == 0 and i % 7 == 0: # compruebo si el número es divisible entre 5 y 7
        print(i)
```

## Ejercicio 8
Implementa un programa que imprima por pantalla el siguiente patrón

5 4 3 2 1 

4 3 2 1 

3 2 1 

2 1 

1

NOTA: NO hay lineas en blanco entre una línea y otra.

```python
''' Voy a crear un rango descendente de 5 a 1.
Cada valor del rango quiero que sea el número desde
el que empieza el patrón del enunciado
'''

patron = range(5, 0, -1) # voy hacia atrás porque es descendiente

for num in patron: # recorro cada núm del rango
    for i in range(num, 0, -1): # hay que hacer un segundo bucle que me genere la longitud correcta de cada línea
        print(i, end=" ") # imprime los elementos en la misma línea, pero con un espacio entre ellos
    print() # esto crea un salto de línea
    
```

## Ejercicio 9
En este ejercicio vamos a crear un pequeño juego. Se trata de intentar adivinar un numero del 1 al 5. Tenemos dos intentos para acertar. Pasos a seguir:

1. Ya viene implementado con la librería `random` cómo obtener un número aleatorio del 1 al 5
2. Tendrás que declarar en una variable el numero de vidas, y mediante un bucle while, comprobar que todavia quedan vidas.
3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas.
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose".

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas.

```python
import random
```

```python
vidas = 2 
num_random = random.randint(1, 5) # pido al ordenador un número random del 1 al 5

while vidas > 0: # mientras queden vidas, el juego sigue y continúo pidiendo intentos al usuario
    intento = int(input("Adivina el número del 1 al 5 que estoy pensando:")) # hay que convertirlo en entero
    if intento == num_random: # si coinciden en introducido y el random, gana
        print("You win")
        break # si lo acierta nos salimos del bucle
    else:
        vidas -= 1 # si no acierta quitamos una vida
        print("Incorrecto. Te quedan", vidas, "vidas.") # si seacaban las vidas, se ejecuta el else del while
else:
    print("You lose. El número era:", num_random) # esto solo se ejecuta si no se usa el break (si no tiene vidas o si falla)
```
