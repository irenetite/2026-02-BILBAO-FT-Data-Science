## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math

print("Hola, Introduce los grados:")
grados = float(input("Hola, introduce los grados:"))
radianes = float(grados) * (math.pi / 180)
print(f"{grados} grados equivalen a {radianes}.")


# opción 2:
print("Hola, Introduce los grados:")
grados = input("Hola, introduce los grados:")
grados = grados.replace(",", ".")
grados = float(grados)
radianes = float(grados) * (math.pi / 180)
print(f"{grados} grados equivalen a {radianes}.")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas

3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
