## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math

print("Hola, Introduce los grados:")
grados = float(input("Hola, introduce los grados:"))
radianes = float(grados) * (math.pi / 180)
print(f"{grados} grados equivalen a {radianes}.")
````

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

````python
print("Hola, soy tu asistente de cálculo de áreas de paralelogramos.")
base = float(input("Introduce el valor de la base (en metros):"))
print("Introduce el valor de la base (en metros):", base)
altura = float(input("Ahora el de la altura (en metros):"))
print("Ahora el de la altura (en metros):", altura)

area = base * altura
print(f"El área de tu paralelogramo es de {area} metros cuadrados.")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
bisbal = "Lo primero de todo, cómo están los máquinas"
# 1:
print(bisbal.upper())

# 2:
print(bisbal.lower())

# 3:
print(bisbal.title())

# 4:
bisbal_dividido = bisbal.split(' ')
print(bisbal_dividido)

# 5:
print(bisbal.replace(",",";"))

# 6:
print(bisbal.replace("a",""))
print(bisbal.replace("á",""))
````
