## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
# JUEGO DEL AHORCADO

''' Primero quiero ver cómo le pido a python que elija una palabra aleatoria del español.
Le doy una lista de posibles palabras que puede elegir y añado la función 'random.choice()'
después de importar la biblioteca random'''
import random

palabras = ["bosque", "invierno", "luciernaga", "mariposa", "murcielago", \
            "ventana", "cascada", "botella", "cielo", "aurora", "bicho"]
palabra_azar = random.choice(palabras)

# no la printeo porque no quiero que la vea el jugador


# ahora voy a lo gordo, voy a crear una función que devuelva el juego del ahorcado:


def ahorcado(palabra_azar):
    rayas = ["_"]*len(palabra_azar) # lo pongo en una lista porque luego tendré que operar por índice
    num_errores = [] # en esta variable contabilizaré los errores que cometa el usuario
    perdiste = False # creo aquí esta variable que luego necesitaré cuando vaya a hacer un if else dentro del while (en un print final)
    print(" ".join(rayas)) # imprimo las rayas por primera vez para que el ususario vea el número de letras de la palabra
    while "_" in rayas: # mientras haya huecos para adivinar la letra:
        letra_usuario = input("Escribe una letra:") # pido al usuario una letra. al principio lo había puesto fuera del while, pero luego me di cuenta de que solo pedía una letra una vez
        letra_usuario = letra_usuario.lower() # para evitar errores puse que convierta las mayúsculas en minúsculas
        acierto = False # creo una variable de control porque, al parecer simplifica el código y se evitan errores y confusiones. esto lo encontré después de buscar por qué me salían algunos errores que no entendía ni encontraba el origen
        if letra_usuario in num_errores or letra_usuario in rayas: # esto lo puse para que salga un mensaje si repites la misma letra
            print("Ya probaste esa letra.")
            continue
        if len(letra_usuario) != 1 or not letra_usuario.isalpha(): # esto lo añadí porque pensé en el posible error de que alguien metiera más letras, así que, buscando por internet, encontré esta forma con la función .isalpha() -> sirve para comprobar si todos los caracteres son letras

            print("Introduce solo una letra.")
            continue
        for i, letra in enumerate(palabra_azar): # pido que, si acierta una letra de la palabra al azar, sustituya esa raya por la letra adivinada en la posición correspondiente a esa letra
            if letra == letra_usuario:
                rayas[i] = letra
                acierto = True
        if acierto:
            print(" ".join(rayas))
        else: # si el usuario mete una letra que no está en la palabra al azar dará error
            print("Esa letra no está.")
            num_errores.append(letra_usuario)
            if len(num_errores) == 6:
                print("Perdiste, fin del juego.")
                perdiste = True # meto esto porque tuve lío al poner el print de ¡Ganaste! y me salía aunque perdiera, pero añadiendo el if else al final ya me funciona
                break
    if perdiste:
        print(f"La palabra era: {palabra_azar}. Cometiste {len(num_errores)} errores: {num_errores}.")
    else:
        print("¡Ganaste!")
        print(f"La palabra era: {palabra_azar}. Cometiste {len(num_errores)} errores: {num_errores}.")
````
