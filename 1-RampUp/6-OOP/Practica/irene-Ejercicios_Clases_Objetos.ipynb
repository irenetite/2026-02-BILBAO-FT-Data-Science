El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

```python
# 1:

class Tienda():
    # atributos comunes:
    tipo = "electrodomesticos"
    abierta = True
    # atributos propios de cada tienda:
    def __init__(self, nombre:str, direccion:str, num_empleados:int, ventas_ult_3_meses:list):
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_ult_3_meses = ventas_ult_3_meses

    # metodo calcula ventas totales:
    def ventas_totales(self) -> float:  # le digo que me devuelva un float, que, por defecto, me había puesto int
        return sum(self.ventas_ult_3_meses)
    # metodo que calcula la media de las ventas por empleado:
    def media_ventas_empleado(self):
        return self.ventas_totales() / self.num_empleados
    # metodo que devuelve en un string el nombre de la tienda, junto con su dirección:
    def string_nombre_direcc(self):
        return self.nombre + self.direccion
    # metodo que obtiene las ventas del último mes.
    def ventas_ult_mes(self):
        return self.ventas_ult_3_meses[-1]
    '''metodo para dar una proyección de las ventas en caso de haber invertido X dinero en marketing
    en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos
    3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de
    (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve
    las ventas con el aumento:'''
    ''' CREO QUE MAL O MÁS COMPLICAO
    def proyeccion_marketing(self, x):
        if x < 1000:
            factor = 1.2
        else:
            factor = 1.5
            
        self.ventas_ult_3_meses = [v * factor for v in self.ventas_ult_3_meses] # hay que reescribir las ventas
        return self.ventas_ult_3_meses
    '''
    
    def proyeccion_ventas(self, inversion):
        if inversion < 1000:
            for i, mes in enumerate(self._ult_3_meses):
                self._ult_3_meses[i] = mes*1.2
        elif inversion >= 1000:
            for i, mes in enumerate(self._ult_3_meses):
                self._ult_3_meses[i] = mes*1.5
        return self._ult_3_meses
                        
                        
                        
```

```python
# 2. Crear tres tiendas con datos inventados:

t_bilbao = Tienda("Bilbao", "Iparraguirre kalea", 20, [120000, 40000, 60000])

t_leioa = Tienda("Leioa", "Errekalde kalea", 6, [6000, 8000, 6000])

t_barakaldo = Tienda("Barakaldo", "Gordoniz kalea", 12, [15000, 18000, 16000])
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    # atributos comunes:
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0 # por defecto están parados

    def __init__(self, raza:str, pelo = "marron", dueno = False):
        '''atributos específicos: raza, pelo marrón por defecto y sin dueño por defecto'''
        self.raza = raza
        self.pelo = pelo
        self.dueno = dueno
    
    def andar(self, aumento_velocidad): 
        '''metodo andar con argumento de entrada 'aumento_velocidad'.
        Este valor se le sumará a la velocidad que ya llevaba el perro.'''
        self.velocidad = aumento_velocidad + self.velocidad
        return self.velocidad # el return lo pongo abajo porque los return no pueden llevar asignaciones
    
    def parar(self):
        '''metodo parar con velocidad a 0
        '''
        self.velocidad = 0 # asigno la velocidad a 0
        return self.velocidad
    
    def ladrar(self, arg):
        ''' metodo ladrar con argumento de entrada el str GUAU!+argumento de entrada'''
        return "GUAU! " + arg # añado un espacio para que me separe el GUAU! del mensaje
```

```python
# objeto de tipo Perro, sin dueño
perro_de_pablo = Perro(raza = "teckel")
print(perro_de_pablo.__dict__)

# compruebo atributos
print(perro_de_pablo.patas)
print(perro_de_pablo.orejas)
print(perro_de_pablo.ojos)
print(perro_de_pablo.velocidad)
print(perro_de_pablo.raza)
print(perro_de_pablo.pelo)
print(perro_de_pablo.dueno)
```

```python
# hago que el perro ande
perro_de_pablo.andar(5)
print(perro_de_pablo.velocidad)

perro_de_pablo.andar(3)
print(perro_de_pablo.velocidad)
```

```python
# hago que el perro pare
perro_de_pablo.parar()
print(perro_de_pablo.velocidad)
```

```python
# hago que ladre
print(perro_de_pablo.ladrar("Quiero jugar!"))  
```
