## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def num_a_dia(num):
    dias = {1: "Lunes",
            2: "Martes",
            3: "Miércoles",
            4: "Jueves",
            5: "Viernes",
            6: "Sábado",
            7: "Domingo"}
    return dias.get(num, "Número fuera de rango")
````

````python
# pruebas:
print(num_a_dia(4))
print(num_a_dia(7))
print(num_a_dia(3))
````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comparacion(n1, n2):
    if n1 == n2:
        return "Son iguales"
    if n1 > n2:
        return f"{n1} es mayor que {n2}"
    if n1 < n2:
        return f"{n1} es menor que {n2}"
````

````python
# pruebas:
print(comparacion(2,2))
print()
print(comparacion(1,2))
print()
print(comparacion(3,2))
````

````python
# danillo

def comparar (a,b):
    match (a,b):
        case (a,b) if a==b:
            print(f"{a} = {b}")
        case (a,b) if a>b:
            print(f"{a} mayor que {b}")
        case (a,b) if a<b:
            print(f"{a} menor que {b}")
            
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def contar_letra(texto, letra):
    texto = texto.lower()
    letra = letra.lower()
    contador = 0
    for caracter in texto:
        if caracter == letra:
            contador += 1
    return contador


'''
igual también es válida esta forma más abreviada:

def contar_letra(texto, letra):
    return texto.lower().count(letra.lower())

'''
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
# encapsular en una función
def contador_letras(palabra):
    return {letra: palabra.count(letra) for letra in set(palabra)}
````

````python
# pruebas:
print(contador_letras("hola"))
print(contador_letras("irene"))
print(contador_letras("pluscoamperfecto"))
````

````python
# danillo:
def gendict(string):
    d = {}
    for letra in string:
        if letra in d.keys():
            d[letra] = d[letra]+1 # cada vez q pases por la letra suma 1 a la cantidad de veces q pasas
        else:
            d[letra] = 0
            d[letra] += 1
    return d
````

````python
gendict("Hola")
````

````python
gendict("paranguiracutimiria")
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def anad_elim_lista(lista, comando, elemento=None):
    if comando == "add":
        if elemento is not None:
            lista.append(elemento)
        return lista
    elif comando == "remove":
        if elemento in lista:
            lista.remove(elemento)
        return lista
    else:
        return "Ese comando no es válido"
````

````python
# pruebas:
lista_prueba = [1, 2, 3]

print(anad_elim_lista(lista_prueba, "add", 4))

print(anad_elim_lista(lista_prueba, "remove", 2))

print(anad_elim_lista(lista_prueba, "remove", 10))

print(anad_elim_lista(lista_prueba, "hola"))
````
