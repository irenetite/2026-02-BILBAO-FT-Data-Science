## 2. Enunciados de los Ejercicios

1. Obtener el nombre de los empleados que hayan repartido al bar **Stop** durante la semana del 17 al 23 de octubre de 2005.
2. Obtener el Cif y nombre de los bares a los que se ha repartido cerveza de tipo **Botella** y capacidad inferior a 1 litro, ordenados por localidad.
3. Obtener los repartos (nombre del bar, envase y capacidad de la bebida, fecha y cantidad) realizados por **Prudencio Caminero**.
4. Obtener los bares a los que se les ha repartido envases de tipo **botella** y capacidad 0.2 ó 0.33.
5. Nombre de los empleados que han repartido a los bares **"Stop"** y **"Las Vegas"** cervezas con envase botella. 
6. Obtener el nombre y número de viajes que ha realizado cada empleado fuera de **Villa Botijo**.
7. Obtener el nombre y localidad del bar que más litros de cerveza ha comprado.
8. Obtener los bares que han adquirido todos los tipos de cerveza con envase de botella y capacidad menor que 1 litro.
9. Subir un 5% el sueldo del empleado que más días haya trabajado.
10. Insertar un nuevo reparto del empleado **"Vicente Merario"** al bar **"Stop"** de 48 cervezas de tipo lata el día 10/26/05.

```python
import pandas as pd
import sqlite3

# creo conexión a base de datos
connection = sqlite3.connect(":memory:")

# cursor
crsr = connection.cursor()
```

```python
# creo las tablas

# tabla CERVEZAS
# CodC: cod birra (PK)
# Envase
# Capacidad
# Stock
crsr.execute('''
CREATE TABLE CERVEZAS (
    CodCerv TEXT PRIMARY KEY,
    Envase TEXT,
    Capacidad REAL,
    Stock INTEGER
)
''')

# tabla BARES
# CodB (PK)
# Cif: código de identificación fiscal
# Nombre: nombre del bar
# Localidad: localidad donde se encuentra
crsr.execute('''
CREATE TABLE BARES (
    CodBar TEXT PRIMARY KEY,
    Cif TEXT,
    Nombre TEXT,
    Localidad TEXT
)
''')

# EMPLEADOS
# CodEmp (PK)
# Nombre
# Sueldo
crsr.execute('''
CREATE TABLE EMPLEADOS (
    CodEmp INTEGER PRIMARY KEY,
    Nombre TEXT,
    Sueldo REAL
)
''')

# REPARTO
# CodEmp (FK)
# CodBar (FK)
# CodCerv ( FK)
# Fecha
# Cantidad
crsr.execute('''
CREATE TABLE REPARTO (
    CodEmp INTEGER,
    CodBar TEXT,
    CodCerv TEXT,
    Fecha TEXT,
    Cantidad INTEGER,
    FOREIGN KEY (CodEmp) REFERENCES EMPLEADOS(CodEmp),
    FOREIGN KEY (CodBar) REFERENCES BARES(CodBar),
    FOREIGN KEY (CodCerv) REFERENCES CERVEZAS(CodCerv)
)
''')
```

```python
# inserto datos en CERVEZAS
cervezas_data = [('01', 'Botella', 0.2, 3600),
    ('02', 'Botella', 0.33, 1200),
    ('03', 'Lata', 0.33, 2400),
    ('04', 'Botella', 1.0, 288),
    ('05', 'Barril', 60.0, 30)]
crsr.executemany('INSERT INTO CERVEZAS VALUES (?, ?, ?, ?)', cervezas_data)

# inserto datos enBARES
bares_data = [('001', '11111111X', 'Stop', 'Villa Botijo'),
    ('002', '22222222Y', 'Las Vegas', 'Villa Botijo'),
    ('003', '-', 'Club Social', 'Las Ranas'),
    ('004', '33333333Z', 'Otra Ronda', 'La Esponja')]
crsr.executemany('INSERT INTO BARES VALUES (?, ?, ?, ?)', bares_data)

# inserto datos en EMPLEADOS
empleados_data = [(1, 'Prudencio Caminero', 120000),
    (2, 'Vicente Merario', 110000),
    (3, 'Valentin Siempre', 100000)]
crsr.executemany('INSERT INTO EMPLEADOS VALUES (?, ?, ?)', empleados_data)

# inserto datos en REPARTO
reparto_data = [(1, '001', '01', '10/21/05', 240),
    (1, '001', '02', '10/21/05', 48),
    (1, '002', '03', '10/22/05', 60),
    (1, '004', '05', '10/22/05', 4),
    (2, '002', '03', '10/22/05', 48),
    (2, '002', '05', '10/23/05', 2),
    (2, '004', '01', '10/23/05', 480),
    (2, '004', '02', '10/24/05', 72),
    (3, '003', '03', '10/24/05', 48),
    (3, '003', '04', '10/25/05', 20)]
crsr.executemany('INSERT INTO REPARTO VALUES (?, ?, ?, ?, ?)', reparto_data)

# guardo cambios enbase de datos
connection.commit()
```

```python
# creo función para ejecutar consultas SQL y obtener resultados como df
# vemos los resultados de forma tabular
def sql_query(query):
    '''
    ejecuta una consulta SQL y devuelve los resultados como un df de pandas
    parámetros: query (consulta SQL a ejecutar)
    deveulve: pd.DataFrame (resultados de la consulta con nomb de cols)
    '''
    crsr.execute(query) # ejecuto consulta SQL
    
    ans = crsr.fetchall() # obtengo todos los resultados
    
    names = [description[0] for description in crsr.description] # obtengo nomb de las cols desde la descripción del cursor
    
    return pd.DataFrame(ans, columns=names) # creo y devuelvo el df
```

```python
# ejercicio 1

# uno tablas EMPLEADOS, REPARTO y BARES, filtro por por Nombre = 'Stop' y por el rango de fechas
# fechas en formato MM/DD/YY, así que busco entre '10/17/05' y '10/23/05'
query1 = '''
SELECT DISTINCT e.Nombre
FROM EMPLEADOS e
JOIN REPARTO r ON e.CodEmp = r.CodEmp
JOIN BARES b ON r.CodBar = b.CodBar
WHERE b.Nombre = 'Stop'
  AND r.Fecha >= '10/17/05'
  AND r.Fecha <= '10/23/05'
'''

result1 = sql_query(query1)
display(result1)
```

```python
# ejercicio 2 
# uno BARES, REPARTO y CERVEZAS, filtro por tipo de envase 'Botella' y capacidad menor a 1 litro
# uso DISTINCT para evitar duplicados y ordeno por Localidad

query2 = '''
SELECT DISTINCT b.Cif, b.Nombre, b.Localidad
FROM BARES b
JOIN REPARTO r ON b.CodBar = r.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
WHERE c.Envase = 'Botella'
  AND c.Capacidad < 1
ORDER BY b.Localidad
'''

result2 = sql_query(query2)
display(result2)
```

```python
# ejercicio 3

# uno tablasBARES (para nombre del bar), CERVEZAS (para envase y capacidad), REPARTO (para fecha y cantidad)
# filtro por nomb empleado

query3 = '''
SELECT b.Nombre AS Nombre_Bar, c.Envase, c.Capacidad, r.Fecha, r.Cantidad
FROM REPARTO r
JOIN BARES b ON r.CodBar = b.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
JOIN EMPLEADOS e ON r.CodEmp = e.CodEmp
WHERE e.Nombre = 'Prudencio Caminero'
'''

result3 = sql_query(query3)
display(result3)
```

```python
# ejercicio 4

# unno tablas y filtro por tipo de envase y capacidades
query4 = '''
SELECT DISTINCT b.Nombre, b.Localidad
FROM BARES b
JOIN REPARTO r ON b.CodBar = r.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
WHERE c.Envase = 'Botella'
  AND c.Capacidad IN (0.2, 0.33)
'''

result4 = sql_query(query4)
display(result4)
```

```python
# ejercicio 5

# uno INTERSECT para encontrar empleados que han repartido a AMBOS bares
# 1a query: empleados que repartieron a stop con botella
# 2a query: empleados que repartieron a las vegas con botella
query5 = '''
SELECT DISTINCT e.Nombre
FROM EMPLEADOS e
JOIN REPARTO r ON e.CodEmp = r.CodEmp
JOIN BARES b ON r.CodBar = b.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
WHERE b.Nombre = 'Stop'
  AND c.Envase = 'Botella'

INTERSECT

SELECT DISTINCT e.Nombre
FROM EMPLEADOS e
JOIN REPARTO r ON e.CodEmp = r.CodEmp
JOIN BARES b ON r.CodBar = b.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
WHERE b.Nombre = 'Las Vegas'
  AND c.Envase = 'Botella'
'''


result5 = sql_query(query5)
display(result5)
```

```python
# ejercicio6

# uno tablas y filtros bares que NO están en villa botijo, agrupo por empleado y cuentoel número de repartos (viajes)
query6 = '''
SELECT e.Nombre, COUNT(*) AS Num_Viajes
FROM EMPLEADOS e
JOIN REPARTO r ON e.CodEmp = r.CodEmp
JOIN BARES b ON r.CodBar = b.CodBar
WHERE b.Localidad != 'Villa Botijo'
GROUP BY e.CodEmp, e.Nombre
'''



result6 = sql_query(query6)
display(result6)
```

```python
# ejercicio 7

# uno las tablas y calculo los litros totales por bar, ordeno de mayor a menor y tomo solo el primero
query7 = '''
SELECT b.Nombre, b.Localidad, SUM(r.Cantidad * c.Capacidad) AS Litros_Totales
FROM BARES b
JOIN REPARTO r ON b.CodBar = r.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
GROUP BY b.CodBar, b.Nombre, b.Localidad
ORDER BY Litros_Totales DESC
LIMIT 1
'''

result7 = sql_query(query7)
display(result7)
```

```python
# ejercicio 8
# cuentocuántos tipos de cerveza existen con envase =botella y capacidad<1
# para cada bar, cuento cuántos tipos distintos compró
# filtrolo s bares cuya cuenta iguala al total de tipos existentes, eesto asegura que el bar ha comprado TODOS los tipos disponibles
query8 = '''
SELECT b.Nombre, b.Localidad
FROM BARES b
JOIN REPARTO r ON b.CodBar = r.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
WHERE c.Envase = 'Botella'
  AND c.Capacidad < 1
GROUP BY b.CodBar, b.Nombre, b.Localidad
HAVING COUNT(DISTINCT c.CodCerv) = (
    SELECT COUNT(DISTINCT CodCerv)
    FROM CERVEZAS
    WHERE Envase = 'Botella'
      AND Capacidad < 1
)
'''

result8 = sql_query(query8)
display(result8)
```

```python
# ejercicio 9

# identifoco empleado que más días distintos ha trabajado, luego actualizo su sueldo multiplicándolo por 1.05 (aumento del 5%)
# uso una subconsulta para encontrar al empleado con más días trabajados
query9 = '''
UPDATE EMPLEADOS
SET Sueldo = Sueldo * 1.05
WHERE CodEmp = (
    SELECT CodEmp
    FROM (
        SELECT r.CodEmp, COUNT(DISTINCT r.Fecha) AS Dias_Trabajados
        FROM REPARTO r
        GROUP BY r.CodEmp
        ORDER BY Dias_Trabajados DESC
        LIMIT 1
    )
)
'''


crsr.execute(query9)
connection.commit()

query9_verify = '''
SELECT e.CodEmp, e.Nombre, e.Sueldo,
       (SELECT COUNT(DISTINCT r.Fecha) FROM REPARTO r WHERE r.CodEmp = e.CodEmp) AS Dias_Trabajados
FROM EMPLEADOS e
ORDER BY e.CodEmp
'''
result9 = sql_query(query9_verify)
display(result9)
```

```python
# ejercicio10


# insertonuevo reparto usando subconsultas para obtener los códigos
query10 = '''
INSERT INTO REPARTO (CodEmp, CodBar, CodCerv, Fecha, Cantidad)
VALUES (
    (SELECT CodEmp FROM EMPLEADOS WHERE Nombre = 'Vicente Merario'),
    (SELECT CodBar FROM BARES WHERE Nombre = 'Stop'),
    (SELECT CodCerv FROM CERVEZAS WHERE Envase = 'Lata' LIMIT 1),
    '10/26/05',
    48
)
'''

# ejecutinserci´on
crsr.execute(query10)
connection.commit()


# muestr  repartos de vicente merario al bar stop
query10_verify = '''
SELECT e.Nombre AS Empleado,b.Nombre AS Bar,c.Envase,c.Capacidad,r.Fecha,r.Cantidad
FROM REPARTO r
JOIN EMPLEADOS e ON r.CodEmp = e.CodEmp
JOIN BARES b ON r.CodBar = b.CodBar
JOIN CERVEZAS c ON r.CodCerv = c.CodCerv
WHERE e.Nombre = 'Vicente Merario'
  AND b.Nombre = 'Stop'
  AND r.Fecha = '10/26/05';
'''
result10 = sql_query(query10_verify)
display(result10)
```
