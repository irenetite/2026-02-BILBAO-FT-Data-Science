### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
prod_caro = chipo.loc[(chipo['item_price']/chipo['quantity']).idxmax(), 'item_name']
print(prod_caro)
```

### Step 4. How many products cost more than $10.00?

```python
prods_mas_10_dol = chipo[chipo["item_price"] > 10]
len(prods_mas_10_dol)
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
# tengoq eu contar pedidos únicos, no líneas como hice antes
'''en la tarea Occupation usé el método nunique() que encontré para que me devuelva el número de elementos
distintos en un eje especificado y me sirve ahora para este:'''
ped_prod_mas_10 = chipo[chipo["item_price"] > 10]["order_id"].nunique()
ped_prod_mas_10
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
''' pedidos cuyo total > 10, no productos individuales
no es lo mismo, un pedido puede tener productos baratitos, pero sumar más de 10
groupby() agrupa datos para obtener total gastado por pedido y sum() suma los precios
de todos los productos dentro de ese pedido '''
total_por_pedido = chipo.groupby("order_id")["item_price"].sum()
peds_total_mayor_10 = (total_por_pedido > 10).sum() # cuento cuántos pedidos tienen un total superior a 10
peds_total_mayor_10
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
'''busco filas donde item_price > 10 Y, ADEMÁS, para asegurarme de que es un precio
unitario, igualo quantity == 1. uso nunique para contar num de elementos distintos''' 
peds_prod_uno_10 = chipo[(chipo["item_price"] > 10) & (chipo["quantity"] == 1)]["order_id"].nunique()
peds_prod_uno_10
```
