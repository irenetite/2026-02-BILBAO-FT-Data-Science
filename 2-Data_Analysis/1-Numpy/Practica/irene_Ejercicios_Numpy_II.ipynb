## Ejercicio 12
Encuentra los índices de los valores que no sean 0 del siguiente vector [1,2,0,0,4,0]. El output sería [1, 2 4]. [Encuentra el método apropiado en la documentación.](https://numpy.org/doc/) Encapsúlalo en una función, con un argumento de entrada de un vector.

````python
array_12 = np.array([1, 2, 0, 0, 4, 0])
print(array_12)

def indices_no_cero(vector):
    return np.nonzero(vector)[0]


print(indices_no_cero(array_12))
````

````python
# unai
def non_zero(x):
    return x[x != 0]

print(non_zero(np.array([1,2,0,0,4,0])))
````

Crea una matriz de 8x8, y rellénala con 0s y 1s. Tienes que replicar:

## Ejercicio 18
Crea una matriz aleatoria de 5x5 y normaliza sus valores. Para normalizar tienes que restar sus valores por la media(`np.mean`) y dividir el resultado por la desviación estándar(`np.std`).

````python
matriz_18 = np.random.random((5, 5))


media_matriz_18 = np.mean(matriz_18)
desv_matriz_18 = np.std(matriz_18)


matriz_18_norm = (matriz_18 - media_matriz_18) / desv_matriz_18

print("Matriz original:\n", matriz_18)
print("\nMedia:", media_matriz_18)
print("Desviación estándar:", desv_matriz_18)
print("\nMatriz normalizada:\n", matriz_18_norm)
````

````python
# unai
mat = np.random.randint(0,5, (5,5))

print((mat - mat.mean())/mat.std())
````

## Ejercicio 19
1. Crea una secuencia de números que vayan del 0 al 10.
2. Crea un nuevo array, que parta del anterior, pero sustituya por el valor -1, todos los elementos cuyo valor se encuentre entre el 3 y el 8.
3. Crea un nuevo array que niegue todos los elementos entre el 3 y el 8

````python
array_19 = np.arange(0, 11)
print(array_19)

array_19_mod = array_19.copy()
array_19_mod[(array_19_mod >= 3) & (array_19_mod <= 8)] = -1
print(array_19_mod)


array_19_neg = array_19.copy()
array_19_neg[(array_19_neg >= 3) & (array_19_neg <= 8)] *= -1
print(array_19_neg)
````

````python
# fátima
seq = np.arange(11)
array_19_a = np.where((seq > 3) & (seq < 8), -1, seq)
print(array_19_a)

array_19_b = np.where((seq > 3) & (seq < 8), -seq, seq)
print(array_19_b)
````
