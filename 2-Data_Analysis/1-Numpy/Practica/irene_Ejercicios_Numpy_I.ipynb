## Ejercicio 8

```Python

```

````python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1,10).reshape(2,-1)

array_concat_8 = np.vstack((a, b))
print(array_concat_8)
````

## Ejercicio 10
Encuentra todos los elementos en común entre ambos arrays. [Para ello usa la función `intersect1d`](https://numpy.org/doc/stable/reference/generated/numpy.intersect1d.html)

```Python
a = np.array([1,2,3,2,3,4,3,4,5,6])
```

## Ejercicio 20
Dado el siguiente array

```Python
x = np.random.randint(10, 20, size = (5,2,6))
```

1. ¿Cuántos elementos tiene el array?
2. ¿Cuántas dimensiones?
3. ¿Cuánta es la longitud cada dimensión?

````python
x = np.random.randint(10, 20, size = (5,2,6))

# 1:
print(x.size)
# 5*2*6 = 60

# 2:
print(x.ndim)

# 3:
print(x.shape)
# la longitud primera dimensión: 5
# la longitud segunda dimensión: 2
# la longitud tercera dimensión: 6
````
