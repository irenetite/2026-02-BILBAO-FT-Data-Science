# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = '''
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
'''
```

Ejemplo:
* @thebridgeschool.tech
* @uc3m.es
* @mi-correo.net
* @thebridgeschool.es

```python
# voy a crear el patrón para extraer dominios de emails

# sería @ seguido de caracteres alfanuméricos, guiones y puntos hasta el final de la línea
pattern = r'@[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}'

# extraigo todos los dominios que coincidan
domains = re.findall(pattern, emails)

# ahora mediente bucle
for domain in domains:
    print(domain)
```
