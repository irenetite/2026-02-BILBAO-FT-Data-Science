# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = '''
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
'''
```

Ejemplo:
* @thebridgeschool.tech
* @uc3m.es
* @mi-correo.net
* @thebridgeschool.es

```python

# [a-zA-Z0-9.-]+ coincide con el nombre de dominio (letras, dígitos, puntos, guiones)
# \. coincide con el punto antes
# [a-zA-Z]{2,} al menos 2 letras
pattern = r'@[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}'
re.findall(pattern, emails)
```
