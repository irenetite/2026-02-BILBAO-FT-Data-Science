## 6. Obtener el Mapeo de Géneros

### ¿Por qué necesitamos un mapeo de géneros?

Cuando buscamos películas en TMDB, los géneros vienen como **IDs numéricos** (ej: `[28, 12, 14]`), no como nombres. Necesitamos:

1. Obtener la lista completa de géneros con sus IDs
2. Crear un diccionario para convertir IDs → Nombres
3. Aplicar este mapeo a nuestras películas

### Endpoint: Genre List

```
GET /genre/movie/list
```

### Parámetros:

| Parámetro | Valor | Descripción |
|-----------|-------|-------------|
| api_key | TU_API_KEY | Tu clave de API |
| language | es-ES | Idioma de los géneros |

### Respuesta esperada:

```json
{
  "genres": [
    {"id": 28, "name": "Acción"},
    {"id": 12, "name": "Aventura"},
    {"id": 14, "name": "Fantasía"},
    ...
  ]
}
```

---

````python
# ============================================================================
# FUNCIÓN: OBTENER MAPEO DE GÉNEROS
# ============================================================================

def get_genres_map(api_key, language="es-ES"):
    """
    Obtiene el diccionario de géneros de películas desde la API de TMDB.
    
    Parámetros:
    - api_key: Tu clave de API de TMDB
    - language: Idioma de los géneros (por defecto: español)
    
    Retorna:
    - Un diccionario {id_genero: nombre_genero}
    """
    
    # Endpoint para obtener la lista de géneros
    endpoint = "/genre/movie/list"
    
    # Parámetros de la petición
    params = {
        "api_key": api_key,
        "language": language
    }
    
    # Hacemos la petición GET a la API
    respuesta = requests.get(url_base + endpoint, params=params)
    
    # Verificamos que la petición fue exitosa
    if respuesta.status_code == 200:
        datos = respuesta.json()
        
        # Creamos el diccionario: {id: nombre}
        # Usamos un diccionario por comprensión
        generos_map = {genero["id"]: genero["name"] for genero in datos["genres"]}
        
        print(f"✅ Se obtuvieron {len(generos_map)} géneros disponibles.")
        return generos_map
    else:
        print(f"❌ Error al obtener géneros. Código: {respuesta.status_code}")
        print(f"   Mensaje: {respuesta.text}")
        return {}

# ============================================================================
# EJECUTAR LA FUNCIÓN
# ============================================================================

# Llamamos a la función y guardamos el resultado
diccionario_generos = get_genres_map(api_key, idioma)

# Mostramos algunos géneros como ejemplo
print("\n📋 Algunos géneros disponibles:")
for i, (id_gen, nombre_gen) in enumerate(diccionario_generos.items()):
    if i < 10:  # Mostramos solo los primeros 10
        print(f"   ID {id_gen}: {nombre_gen}")
print("   ...")
````

## 7. Búsqueda de Películas

### Endpoint: Search Movie

```
GET /search/movie
```

### Parámetros:

| Parámetro | Valor | Descripción |
|-----------|-------|-------------|
| api_key | TU_API_KEY | Tu clave de API |
| query | "M" | Término de búsqueda |
| language | es-ES | Idioma de los resultados |
| page | 1 | Número de página |

### Paginación

La API de TMDB devuelve **máximo 20 resultados por página**. Para obtener más películas, necesitamos:

1. Hacer múltiples peticiones cambiando el parámetro `page`
2. Concatenar todos los resultados
3. Detenernos cuando no haya más resultados

### Respuesta esperada:

```json
{
  "page": 1,
  "total_pages": 50,
  "total_results": 1000,
  "results": [
    {
      "id": 123,
      "title": "Matrix",
      "release_date": "1999-03-30",
      "genre_ids": [28, 878],
      "vote_average": 8.2,
      "overview": "Una historia de ciencia ficción..."
    },
    ...
  ]
}
```

---

````python
# ============================================================================
# FUNCIÓN: BUSCAR PELÍCULAS
# ============================================================================

def search_movies(api_key, query, language="es-ES", max_pages=5):
    """
    Busca películas por título en la API de TMDB.
    
    Parámetros:
    - api_key: Tu clave de API de TMDB
    - query: Término de búsqueda (ej: inicial del nombre)
    - language: Idioma de los resultados (por defecto: español)
    - max_pages: Número máximo de páginas a obtener (por defecto: 5)
    
    Retorna:
    - Una lista de diccionarios con la información de las películas
    """
    
    # Endpoint para buscar películas
    endpoint = "/search/movie"
    
    # Lista para guardar todas las películas encontradas
    todas_peliculas = []
    
    print(f"🔍 Buscando películas que empiezan por '{query}'...")
    print()
    
    # Bucle para obtener múltiples páginas de resultados
    for pagina in range(1, max_pages + 1):
        
        # Parámetros de la petición
        params = {
            "api_key": api_key,
            "query": query,
            "language": language,
            "page": pagina
        }
        
        # Hacemos la petición GET a la API
        respuesta = requests.get(url_base + endpoint, params=params)
        
        # Verificamos que la petición fue exitosa
        if respuesta.status_code == 200:
            datos = respuesta.json()
            
            # Obtenemos los resultados de esta página
            resultados_pagina = datos.get("results", [])
            
            # Si no hay más resultados, salimos del bucle
            if not resultados_pagina:
                print(f"   Página {pagina}: No hay más resultados.")
                break
            
            # Añadimos las películas a nuestra lista
            todas_peliculas.extend(resultados_pagina)
            
            print(f"   Página {pagina}: {len(resultados_pagina)} películas encontradas.")
            
            # Si llegamos a la última página, salimos del bucle
            if pagina >= datos.get("total_pages", 1):
                break
        else:
            print(f"❌ Error en la página {pagina}. Código: {respuesta.status_code}")
            break
    
    print()
    print(f"✅ Total de películas obtenidas: {len(todas_peliculas)}")
    return todas_peliculas

# ============================================================================
# EJECUTAR LA BÚSQUEDA
# ============================================================================

# Buscamos películas que empiecen por nuestra inicial
# max_pages=3 obtendrá hasta 60 películas (20 por página)
lista_peliculas = search_movies(api_key, mi_inicial, idioma, max_pages=3)

# Mostramos las primeras 3 películas como ejemplo
if lista_peliculas:
    print("\n📋 Ejemplo de las primeras películas:")
    for i, pelicula in enumerate(lista_peliculas[:3]):
        print(f"\n   {i+1}. {pelicula.get('title', 'N/A')}")
        print(f"      Fecha: {pelicula.get('release_date', 'N/A')}")
        print(f"      Géneros (IDs): {pelicula.get('genre_ids', [])}")
        print(f"      Puntuación: {pelicula.get('vote_average', 'N/A')}")
````

## 8. Procesamiento de Datos

### Transformación de géneros

Ahora necesitamos convertir los **IDs de géneros** a **nombres reales**. Por ejemplo:

```
Antes: [28, 12, 14]
Después: "Acción, Aventura, Fantasía"
```

### Función de mapeo

Crearemos una función que:

1. Recibe una lista de IDs de géneros
2. Busca cada ID en el diccionario de géneros
3. Une los nombres con comas

### Manejo de valores faltantes

Algunas películas pueden tener:

- Géneros vacíos `[]`
- Fechas de estreno vacías `""`
- Sinopsis vacías `""`

Debemos manejar estos casos apropiadamente.

---

````python
# ============================================================================
# FUNCIÓN: CONVERTIR IDs DE GÉNEROS A NOMBRES
# ============================================================================

def convertir_generos(ids_generos, diccionario_generos):
    """
    Convierte una lista de IDs de géneros a una cadena de texto con los nombres.
    
    Parámetros:
    - ids_generos: Lista de IDs numéricos (ej: [28, 12, 14])
    - diccionario_generos: Diccionario {id: nombre}
    
    Retorna:
    - Cadena de texto con los nombres separados por comas (ej: "Acción, Aventura, Fantasía")
    """
    
    # Si la lista está vacía o es None, devolvemos "N/A"
    if not ids_generos:
        return "N/A"
    
    # Convertimos cada ID a su nombre correspondiente
    # Usamos .get() para manejar IDs que no estén en el diccionario
    nombres_generos = [diccionario_generos.get(id_gen, f"Desconocido ({id_gen})") 
                       for id_gen in ids_generos]
    
    # Unimos los nombres con ", " y devolvemos el resultado
    return ", ".join(nombres_generos)

# ============================================================================
# PROCESAR TODAS LAS PELÍCULAS
# ============================================================================

# Lista para guardar las películas procesadas
peliculas_procesadas = []

print("🔄 Procesando películas...")
print()

for pelicula in lista_peliculas:
    
    # Extraemos cada campo que nos interesa
    pelicula_procesada = {
        "id": pelicula.get("id", "N/A"),
        "title": pelicula.get("title", "N/A"),
        "release_date": pelicula.get("release_date", "N/A"),
        "genres": convertir_generos(pelicula.get("genre_ids", []), diccionario_generos),
        "vote_average": pelicula.get("vote_average", "N/A"),
        "overview": pelicula.get("overview", "N/A")
    }
    
    # Añadimos la película procesada a la lista
    peliculas_procesadas.append(pelicula_procesada)

print(f"✅ {len(peliculas_procesadas)} películas procesadas correctamente.")
print()

# Mostramos un ejemplo del resultado
if peliculas_procesadas:
    print("📋 Ejemplo de película procesada:")
    print(peliculas_procesadas[0])
````

````python
# ============================================================================
# CREACIÓN DEL DATAFRAME
# ============================================================================

# Creamos el DataFrame a partir de la lista de películas procesadas
# pandas convierte automáticamente la lista de diccionarios en una tabla
df_peliculas = pd.DataFrame(peliculas_procesadas)

# ============================================================================
# VISUALIZACIÓN DEL DATAFRAME
# ============================================================================

# .head() muestra las primeras 5 filas del DataFrame
print("📋 Primeras filas del DataFrame:")
print()
print(df_peliculas.head())
print()

# .tail() muestra las últimas 5 filas
print("📋 Últimas filas del DataFrame:")
print()
print(df_peliculas.tail())
print()

# ============================================================================
# INFORMACIÓN TÉCNICA DEL DATAFRAME
# ============================================================================

# .info() muestra información sobre el DataFrame:
# - Número de entradas (filas)
# - Nombres de columnas
# - Tipo de dato de cada columna
# - Uso de memoria
print("ℹ️  Información del DataFrame:")
print()
df_peliculas.info()
print()

# Mostramos las dimensiones del DataFrame
print(f"📐 Dimensiones del DataFrame: {df_peliculas.shape[0]} filas × {df_peliculas.shape[1]} columnas")
print()

# ============================================================================
# ESTADÍSTICAS DESCRIPTIVAS
# ============================================================================

# .describe() muestra estadísticas de las columnas numéricas
print("📊 Estadísticas de la puntuación (vote_average):")
print()
print(df_peliculas["vote_average"].describe())
````

## 10. Exportar a CSV

### ¿Qué es un archivo CSV?

**CSV** (Comma-Separated Values) es un formato de archivo de texto simple que almacena datos tabulares. Cada línea representa una fila y los valores se separan por comas.

### Ejemplo de CSV:

```csv
id,title,release_date,genres,vote_average,overview
123,Matrix,1999-03-30,"Acción, Ciencia Ficción",8.2,"Un hacker descubre..."
456,Inception,2010-07-16,"Acción, Thriller",8.8,"Un ladrón que roba secretos..."
```

### Parámetros de to_csv():

| Parámetro | Valor | Descripción |
|-----------|-------|-------------|
| index | False | No guardar el índice del DataFrame |
| encoding | utf-8 | Codificación para caracteres especiales |

---

````python
# ============================================================================
# EXPORTAR A CSV
# ============================================================================

# Nombre del archivo de salida
# Usamos la inicial para nombrar el archivo
nombre_archivo = f"peliculas_inicial_{mi_inicial}.csv"

# Exportamos el DataFrame a CSV
# index=False significa que no guardamos el índice numérico del DataFrame
# encoding='utf-8-sig' asegura que los caracteres especiales (ñ, acentos) se guarden correctamente
df_peliculas.to_csv(nombre_archivo, index=False, encoding='utf-8-sig')

print(f"✅ Archivo '{nombre_archivo}' exportado correctamente.")
print()

# ============================================================================
# VERIFICAR EL ARCHIVO CREADO
# ============================================================================

# Leemos el archivo CSV para verificar que se creó correctamente
df_verificacion = pd.read_csv(nombre_archivo, encoding='utf-8-sig')

print("📋 Verificación del archivo CSV:")
print(f"   Número de filas: {len(df_verificacion)}")
print(f"   Columnas: {list(df_verificacion.columns)}")
print()

# Mostramos las primeras filas del CSV
print("📋 Primeras filas del CSV:")
print()
print(df_verificacion.head())
````
