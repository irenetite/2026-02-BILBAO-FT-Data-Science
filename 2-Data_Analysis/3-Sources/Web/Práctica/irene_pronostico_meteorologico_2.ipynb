## 2. ¿Qué es una API?

### Definición

**API** significa **Application Programming Interface** (Interfaz de Programación de Aplicaciones). Es un conjunto de reglas y protocolos que permite que dos aplicaciones de software se comuniquen entre sí.

### Analogía sencilla 🍕

Imagina que estás en un restaurante:

| Elemento | Analogía |
|----------|----------|
| **Tú** | El cliente (nuestro programa) |
| **Cocina** | El servidor (la API) |
| **Camarero** | La API (intermediario) |
| **Pedido** | La petición HTTP (request) |
| **Comida** | La respuesta (response) |

### Tipos de peticiones HTTP más comunes:

- **GET**: Obtener datos (lo que usaremos nosotros)
- **POST**: Enviar/crear datos
- **PUT**: Actualizar datos
- **DELETE**: Eliminar datos

## 5. Configuración de Parámetros

### Coordenadas geográficas

Para obtener datos meteorológicos de una ubicación específica, necesitamos proporcionar sus **coordenadas geográficas**:

#### 📍 Latitud
- Mide la distancia angular desde el Ecuador (0°) hasta los polos (90°)
- Positivo = Hemisferio Norte
- Negativo = Hemisferio Sur

#### 📍 Longitud
- Mide la distancia angular desde el Meridiano de Greenwich (0°)
- Positivo = Este
- Negativo = Oeste

### Coordenadas de Bilbao:

| Coordenada | Valor | Significado |
|------------|-------|-------------|
| Latitud | 43.2540 | 43.2540° Norte |
| Longitud | -2.9230 | 2.9230° Oeste |

### Variables

Una **variable** es como una caja donde guardamos información. En este caso:

latitude = 43.2540    # Guardamos el valor 43.2540 en la variable 'latitude'

---

````python
# ============================================================================
# CONFIGURACIÓN DE COORDENADAS GEOGRÁFICAS
# ============================================================================

# Coordenadas de Bilbao, España
# Puedes cambiar estas coordenadas para obtener datos de otra ciudad
# Ejemplo: Madrid (40.4168, -3.7038), Barcelona (41.3851, 2.1734)

latitude = 43.2540    # Latitud de Bilbao (Hemisferio Norte)
longitude = -2.9230   # Longitud de Bilbao (Oeste de Greenwich)

# Mostramos las coordenadas configuradas
print(f"📍 Ubicación configurada:")
print(f"   Latitud:  {latitude}°")
print(f"   Longitud: {longitude}°")
print(f"   Ciudad:   Bilbao, España")
````

## 6. Petición a la API

### Construcción de la URL

Una URL de API tiene varias partes:

```
https://api.open-meteo.com/v1/forecast?latitude=43.2540&longitude=-2.9230&hourly=temperature_2m&forecast_days=7
│                              │
│                              └── Punto final (endpoint): /v1/forecast
└── Dominio base: https://api.open-meteo.com
```

### Parámetros de consulta (query parameters)

Los parámetros se añaden después del `?` y se separan con `&`:

```
?parametro1=valor1&parametro2=valor2&parametro3=valor3
```

### El método GET

Usaremos `requests.get()` que es como decir "por favor, dame los datos de esta URL". Es una petición de **lectura**, no modificamos nada en el servidor.

### Códigos de estado HTTP

| Código | Significado | Descripción |
|--------|-------------|-------------|
| 200 | ✅ OK | La petición fue exitosa |
| 400 | ❌ Bad Request | La petición está mal formada |
| 404 | ❌ Not Found | El recurso no existe |
| 500 | ❌ Internal Server Error | Error del servidor |

---

````python
# ============================================================================
# CONSTRUCCIÓN DE LA URL Y PETICIÓN A LA API
# ============================================================================

# URL base de la API de Open-Meteo
# Esta es la dirección principal a la que haremos la petición
url_base = "https://api.open-meteo.com/v1/forecast"

# Diccionario con los parámetros de la consulta
# Un diccionario en Python es una colección de pares clave-valor
parametros = {
    "latitude": latitude,           # Latitud de Bilbao
    "longitude": longitude,         # Longitud de Bilbao
    "hourly": "temperature_2m",     # Variable: temperatura a 2 metros
    "forecast_days": 7              # Número de días de pronóstico
}

# Mostramos la información de la petición que vamos a hacer
print("🌐 Información de la petición:")
print(f"   URL: {url_base}")
print(f"   Parámetros: {parametros}")
print()

# ============================================================================
# REALIZACIÓN DE LA PETICIÓN HTTP
# ============================================================================

# requests.get() envía una petición GET al servidor
# El servidor procesa nuestra petición y nos devuelve una respuesta
respuesta = requests.get(url_base, params=parametros)

# La respuesta tiene varios atributos importantes:
# - status_code: código de estado HTTP (200 = éxito)
# - text: contenido como texto
# - json(): contenido parseado como JSON

print(f"📡 Estado de la respuesta: {respuesta.status_code}")

# Verificamos si la petición fue exitosa
if respuesta.status_code == 200:
    print("✅ ¡Petición exitosa! Los datos se han obtenido correctamente.")
else:
    print(f"❌ Error en la petición. Código: {respuesta.status_code}")
    print("   Esto podría deberse a:")
    print("   - Problemas de conexión a Internet")
    print("   - La API no está disponible temporalmente")
    print("   - Parámetros incorrectos")
````

## 7. Procesamiento del JSON

### ¿Qué es JSON?

**JSON** (JavaScript Object Notation) es un formato de texto ligero para el intercambio de datos. Es fácil de leer para humanos y máquinas.

### Estructura de JSON

JSON usa dos estructuras principales:

1. **Objetos**: Colecciones de pares clave-valor entre llaves `{}`
2. **Arrays**: Listas ordenadas de valores entre corchetes `[]`

### Ejemplo de la estructura que recibimos:

```json
{
  "latitude": 43.254,
  "longitude": -2.923,
  "hourly": {
    "time": ["2026-03-24T00:00", "2026-03-24T01:00", ...],
    "temperature_2m": [12.5, 12.3, 11.8, ...]
  }
}
```

### Acceder a datos anidados

Para acceder a los datos usamos la notación de corchetes:

```python
data["hourly"]["time"]          # Obtenemos la lista de fechas
data["hourly"]["temperature_2m"] # Obtenemos la lista de temperaturas
```

---

````python
# ============================================================================
# PROCESAMIENTO DE LA RESPUESTA JSON
# ============================================================================

# Convertimos la respuesta a formato JSON (diccionario de Python)
# .json() parsea el texto JSON y lo convierte en estructuras de Python
datos = respuesta.json()

# ============================================================================
# EXPLORACIÓN DE LA ESTRUCTURA DE DATOS
# ============================================================================

# Veamos qué claves principales tiene nuestro JSON
print("📁 Claves principales del JSON:")
for clave in datos.keys():
    print(f"   - {clave}")
print()

# ============================================================================
# EXTRACCIÓN DE LAS TEMPERATURAS Y FECHAS
# ============================================================================

# Accedemos a los datos horarios dentro del JSON
# La estructura es: datos -> hourly -> time / temperature_2m
lista_fechas = datos["hourly"]["time"]
lista_temperaturas = datos["hourly"]["temperature_2m"]

# ============================================================================
# INFORMACIÓN SOBRE LOS DATOS OBTENIDOS
# ============================================================================

print("📊 Resumen de los datos obtenidos:")
print(f"   Número total de registros: {len(lista_temperaturas)} horas")
print(f"   Primera fecha: {lista_fechas[0]}")
print(f"   Última fecha:  {lista_fechas[-1]}")
print()
print(f"🌡️  Estadísticas de temperatura:")
print(f"   Temperatura mínima: {min(lista_temperaturas)}°C")
print(f"   Temperatura máxima: {max(lista_temperaturas)}°C")
print(f"   Temperatura media:  {sum(lista_temperaturas)/len(lista_temperaturas):.2f}°C")
print()

# Mostramos los primeros 5 valores como ejemplo
print("📋 Primeras 5 temperaturas:")
for i in range(5):
    print(f"   {lista_fechas[i]} -> {lista_temperaturas[i]}°C")
````

## 8. Creación del DataFrame

### ¿Qué es un DataFrame?

Un **DataFrame** es una estructura de datos de pandas que representa una **tabla bidimensional** con filas y columnas. Es similar a una hoja de cálculo de Excel o una tabla de base de datos.

### Características:

- Tiene **columnas** con nombres
- Tiene **filas** con índices
- Cada columna puede tener un tipo de dato diferente
- Permite operaciones complejas de forma sencilla

### Creación de un DataFrame

Podemos crear un DataFrame desde un diccionario:

```python
df = pd.DataFrame({
    "Columna1": [valor1, valor2, ...],
    "Columna2": [valor1, valor2, ...]
})
```

### Conversión de fechas

Las fechas en formato texto (string) las convertimos a formato datetime para poder hacer operaciones con ellas:

```python
pd.to_datetime(lista_fechas)  # Convierte strings a objetos datetime
```

---

````python
# ============================================================================
# CREACIÓN DEL DATAFRAME
# ============================================================================

# Creamos un DataFrame con dos columnas:
# - 'Hora': las fechas y horas del pronóstico
# - 'Temperatura': las temperaturas correspondientes

# pd.to_datetime() convierte la lista de strings a objetos datetime
# Esto nos permitirá hacer operaciones con las fechas (filtrar, agrupar, etc.)
df = pd.DataFrame({
    "Hora": pd.to_datetime(lista_fechas),
    "Temperatura": lista_temperaturas
})

# ============================================================================
# VISUALIZACIÓN DEL DATAFRAME
# ============================================================================

# .head() muestra las primeras 5 filas del DataFrame
# Es útil para ver rápidamente cómo están estructurados los datos
print("📋 Primeras filas del DataFrame:")
print(df.head())
print()

# .tail() muestra las últimas 5 filas
print("📋 Últimas filas del DataFrame:")
print(df.tail())
print()

# ============================================================================
# INFORMACIÓN TÉCNICA DEL DATAFRAME
# ============================================================================

# .info() muestra información sobre el DataFrame:
# - Número de entradas (filas)
# - Nombres de columnas
# - Tipo de dato de cada columna
# - Uso de memoria
print("ℹ️  Información del DataFrame:")
df.info()
print()

# Mostramos las dimensiones del DataFrame
print(f"📐 Dimensiones del DataFrame: {df.shape[0]} filas × {df.shape[1]} columnas")
````

## 9. Visualización de Datos

### ¿Por qué visualizar datos?

Las visualizaciones nos permiten:

- 📈 Identificar **patrones** y **tendencias**
- 🔍 Detectar **valores atípicos** (outliers)
- 📊 Comunicar información de forma clara
- 🧠 Comprender grandes volúmenes de datos rápidamente

### Tipo de gráfica: Línea

Para mostrar la evolución de una variable a lo largo del tiempo, la gráfica de **líneas** es la más adecuada porque:

- Muestra la continuidad de los datos
- Permite ver tendencias claramente
- Facilita la comparación de valores

### Elementos de una buena gráfica:

1. **Título**: Describe qué muestra la gráfica
2. **Eje X**: Variable independiente (tiempo)
3. **Eje Y**: Variable dependiente (temperatura)
4. **Leyenda**: Explica qué representa cada línea
5. **Grid**: Ayuda a leer los valores
6. **Etiquetas**: Nombres claros de los ejes

---

````python
# ============================================================================
# CREACIÓN DE LA GRÁFICA DE LÍNEAS
# ============================================================================

# Configuramos el tamaño de la figura
# figsize=(ancho, alto) en pulgadas
# 14x6 es un buen tamaño para ver los detalles
plt.figure(figsize=(14, 6))

# ============================================================================
# DIBUJADO DE LA LÍNEA DE TEMPERATURA
# ============================================================================

# plt.plot() crea la gráfica de líneas
# Parámetros:
#   - df["Hora"]: datos para el eje X (tiempo)
#   - df["Temperatura"]: datos para el eje Y (temperatura)
#   - marker="": no muestra marcadores en cada punto
#   - linewidth=1: grosor de la línea
#   - color="red": color de la línea
#   - label="Temperatura": etiqueta para la leyenda
plt.plot(df["Hora"], 
         df["Temperatura"], 
         marker="", 
         linewidth=1, 
         color="red", 
         label="Temperatura")

# ============================================================================
# TÍTULO Y ETIQUETAS
# ============================================================================

# plt.title() añade el título principal
# fontsize: tamaño de la fuente
# fontweight: grosor de la fuente ("bold" = negrita)
plt.title("🌡️ Pronóstico de Temperaturas en Bilbao (7 días)", 
          fontsize=16, 
          fontweight="bold")

# plt.xlabel() etiqueta del eje X
plt.xlabel("📅 Fecha y Hora", fontsize=12)

# plt.ylabel() etiqueta del eje Y
plt.ylabel("🌡️ Temperatura (°C)", fontsize=12)

# ============================================================================
# FORMATO DE LOS EJES
# ============================================================================

# plt.xticks() configura las marcas del eje X
# rotation=45 rota las etiquetas 45 grados para que no se superpongan
plt.xticks(rotation=45)

# plt.legend() muestra la leyenda con las etiquetas de cada línea
plt.legend()

# plt.grid() añade una cuadrícula de fondo
# alpha=0.3 hace la grid semitransparente (0 = invisible, 1 = opaco)
plt.grid(True, alpha=0.3)

# ============================================================================
# AJUSTE FINAL Y VISUALIZACIÓN
# ============================================================================

# plt.tight_layout() ajusta automáticamente los márgenes
# Esto evita que las etiquetas se corten al guardar o mostrar
plt.tight_layout()

# plt.show() muestra la gráfica
# En Jupyter, esto renderiza la gráfica en la celda de salida
plt.show()

print("✅ Gráfica generada correctamente.")
````

## 10. Estadísticas Descriptivas

### ¿Qué son las estadísticas descriptivas?

Las **estadísticas descriptivas** son medidas numéricas que resumen las características principales de un conjunto de datos.

### Medidas que vamos a obtener:

| Medida | Símbolo | Descripción |
|--------|---------|-------------|
| **count** | n | Número total de observaciones |
| **mean** | μ o x̄ | Media aritmética (promedio) |
| **std** | σ | Desviación estándar (dispersión) |
| **min** | - | Valor mínimo |
| **25%** | Q1 | Primer cuartil (25% de datos por debajo) |
| **50%** | Q2/Mediana | Valor central |
| **75%** | Q3 | Tercer cuartil (75% de datos por debajo) |
| **max** | - | Valor máximo |

### Interpretación:

- **Media**: Temperatura promedio esperada
- **Desviación estándar**: Cuánto varían las temperaturas respecto al promedio
- **Mín/Máx**: Extremos de temperatura en el período
- **Cuartiles**: Distribución de las temperaturas

---

````python
# ============================================================================
# ESTADÍSTICAS DESCRIPTIVAS
# ============================================================================

# .describe() calcula automáticamente las estadísticas principales
# de las columnas numéricas del DataFrame
print("📊 Estadísticas descriptivas de las temperaturas:")
print("=" * 50)
estadisticas = df["Temperatura"].describe()
print(estadisticas)
print("=" * 50)
print()

# ============================================================================
# INTERPRETACIÓN DE LAS ESTADÍSTICAS
# ============================================================================

print("\n📝 Interpretación de resultados:")
print("-" * 50)
print(f"• Número de horas analizadas: {int(estadisticas['count'])}")
print(f"• Temperatura media: {estadisticas['mean']:.2f}°C")
print(f"• Desviación estándar: {estadisticas['std']:.2f}°C")
print(f"• Temperatura mínima: {estadisticas['min']:.1f}°C")
print(f"• Temperatura máxima: {estadisticas['max']:.1f}°C")
print(f"• Mediana (50%): {estadisticas['50%']:.2f}°C")
print(f"• Rango intercuartílico: {estadisticas['25%']:.2f}°C - {estadisticas['75%']:.2f}°C")
print("-" * 50)
print()

# Cálculo adicional: amplitud térmica
amplitud_termica = estadisticas['max'] - estadisticas['min']
print(f"🌡️  Amplitud térmica total: {amplitud_termica:.1f}°C")
print("   (Diferencia entre temperatura máxima y mínima)")
````
