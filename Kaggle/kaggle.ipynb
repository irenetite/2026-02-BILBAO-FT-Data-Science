```python
# Paso 1: Gestionar los identificadores (id y laptop_ID)
# Tal y como comentamos, laptop_ID es ruido y no nos sirve, pero id lo necesitamos para Kaggle
# La mejor práctica es convertir id en el índice del DataFrame y borrar laptop_ID


'''
¿Qué hacemos con id y laptop_ID?
Respuesta corta: Para predecir el precio, no sirven para nada y confundirán a tu modelo.

laptop_ID: Es un identificador interno antiguo del dataset original. Almacena un número que no tiene
ninguna relación lógica con las características físicas ni con el precio del ordenador. Bórrala sin piedad.

id: Este es el identificador de la competición. Tampoco tiene valor predictivo (es ruido para tu algoritmo),
PERO no la borres del todo. Kaggle la va a necesitar en el test.csv para saber a qué ordenador exacto corresponde
la predicción que le envíes en tu sample_submission.csv.

Lo ideal es separar la columna id antes de entrenar o dejarla como el índice (index) del DataFrame para que
el modelo la ignore matemáticamente.

'''
```

```python
# convierto 'id' en el índice del dataframe
df = df.set_index('id')

# elimino 'laptop_ID' porque no aporta valor analítico
df = df.drop(columns=['laptop_ID'])

# muestr5 las primeras filas para comprobarlo
df.head(5)
```

```python
# Paso 2: Limpiar Ram y Weight
# Voy a eliminar las letras para que Pandas reconozca estas columnas como variables estrictamente numéricas.
```

```python
# quito la palabra 'GB' de la RAM y la convierto a número entero
df['Ram'] = df['Ram'].str.replace('GB', '').astype(int)

# quito las letras 'kg' del peso y lo convierto a float
df['Weight'] = df['Weight'].str.replace('kg', '').astype(float)

# compruebo los tipos de datos para asegurar que han cambiado
print(df[['Ram', 'Weight']].dtypes)
```

```python
# muestr5 las primeras filas para comprobarlo
df.head(5)
```

```python
# Paso 3: Extraer información de la Pantalla (ScreenResolution)
# Aquí usamos expresiones regulares para sacar el ancho y el alto en píxeles, y buscamos
# palabras clave para crear columnas de "Sí/No" (1 o 0).
```

```python
# import re # Asegúrate de haber importado 're'

# Extraemos la resolución numérica (ej: de "1920x1080" saca 1920 y 1080)
resolucion = df['ScreenResolution'].str.extract(r'(\d+)x(\d+)')
df['X_res'] = resolucion[0].astype(int)
df['Y_res'] = resolucion[1].astype(int)

# Creamos variables binarias (1 o 0) para características premium
df['Touchscreen'] = df['ScreenResolution'].apply(lambda x: 1 if 'Touchscreen' in x else 0)
df['IPS'] = df['ScreenResolution'].apply(lambda x: 1 if 'IPS' in x else 0)

# Ya podemos borrar la columna de texto original
df = df.drop(columns=['ScreenResolution'])
```

```python
# muestr5 las primeras filas para comprobarlo
df.head(5)
```

```python
# Paso 4: Simplificar Procesador y Gráfica (Cpu y Gpu)
# Los nombres completos son demasiado específicos. Nos interesa la marca y, en el caso
# de la CPU, la velocidad en GHz.
```

```python
# Extraemos la marca (la primera palabra de la cadena de texto)
df['Cpu_Brand'] = df['Cpu'].apply(lambda x: x.split()[0])
df['Gpu_Brand'] = df['Gpu'].apply(lambda x: x.split()[0])

# Extraemos la velocidad del procesador buscando el número decimal antes de "GHz"
df['Cpu_GHz'] = df['Cpu'].str.extract(r'(\d+(?:\.\d+)?)GHz').astype(float)

# Borramos las columnas originales
df = df.drop(columns=['Cpu', 'Gpu'])
```

```python
# muestr5 las primeras filas para comprobarlo
df.head(5)
```

```python
# Paso 5: Desglosar el almacenamiento (Memory)
# Esta es la parte más compleja. Vamos a estandarizar los Terabytes a Gigabytes y a crear
# cuatro columnas separadas dependiendo del tipo de disco duro.
```

```python
# 1. Estandarizar unidades: Convertimos los TB a GB multiplicando por 1000
df['Memory'] = df['Memory'].str.replace(r'(\d+(?:\.\d+)?)TB', 
                                        lambda m: str(int(float(m.group(1))*1000)) + 'GB', 
                                        regex=True)

# 2. Creamos las columnas numéricas inicializadas a 0
df['SSD_GB'] = 0
df['HDD_GB'] = 0
df['Flash_GB'] = 0
df['Hybrid_GB'] = 0

# 3. Rellenamos las columnas iterando por cada fila
for index, row in df.iterrows():
    # Separamos si hay un '+' (ej: "256GB SSD + 1000GB HDD")
    partes = row['Memory'].split('+')
    
    for parte in partes:
        # Buscamos el número de gigas
        match = re.search(r'(\d+)GB', parte)
        if match:
            capacidad = int(match.group(1))
            # Lo sumamos a la columna correspondiente
            if 'SSD' in parte:
                df.at[index, 'SSD_GB'] += capacidad
            elif 'HDD' in parte:
                df.at[index, 'HDD_GB'] += capacidad
            elif 'Flash' in parte:
                df.at[index, 'Flash_GB'] += capacidad
            elif 'Hybrid' in parte:
                df.at[index, 'Hybrid_GB'] += capacidad

# 4. Borramos la columna original
df = df.drop(columns=['Memory'])

# Echamos un vistazo final a cómo ha quedado nuestro dataframe numérico
df.head()
```

```python
# muestr5 las primeras filas para comprobarlo
df.head(5)
```

```python
# Nos quedan unas cuantas columnas categóricas (de texto puro): Company, TypeName, OpSys, Cpu_Brand y Gpu_Brand.

# Para solucionar esto, vamos a usar una técnica llamada One-Hot Encoding (Codificación One-Hot)
# Lo que hace es crear una columna nueva por cada categoría (ej. Company_Apple, Company_Lenovo) y poner un 1 si el
# ordenador es de esa marca y un 0 si no lo es.

# Paso 6: Convertir el texto restante a números (One-Hot Encoding)
# Pandas tiene una función maravillosa llamada get_dummies que hace exactamente esto en una sola línea
# Crea una nueva celda con esto:
```

```python
# Lista de las columnas categóricas que nos quedan
columnas_categoricas = ['Company', 'TypeName', 'OpSys', 'Cpu_Brand', 'Gpu_Brand']

# Aplicamos One-Hot Encoding
# drop_first=True evita el problema de la "multicolinealidad" (una buena práctica en ML)
df = pd.get_dummies(df, columns=columnas_categoricas, drop_first=True)

# ¡Magia! Comprobamos el tamaño de nuestro nuevo dataset
print(f"Ahora nuestro dataset tiene {df.shape[1]} columnas numéricas.")
df.head(2)
```

```python
# Paso 7: Separar la variable a predecir (X e y)
# ¡Ahora sí! El dataset está 100% limpio y numérico. El último paso preparatorio antes de aplicar Machine Learning
# es separar lo que queremos predecir (el precio) de los datos que usaremos para predecirlo (las características).
```

```python
# Lo que nos falta para que sea PERFECTO (Nivel Ganador de Kaggle)
# Para ganar de verdad, hay que pensar como un comprador de portátiles. Si un comprador mira el precio, se fija en
# un par de cosas vitales que nuestra limpieza actual ha pasado un poco por alto. Aquí están las dos killer features
# (características estrella) que deberíamos añadir:

# 1. La Familia del Procesador (Tier)
# En nuestro paso 4, sacamos si el procesador era "Intel" o "AMD", y sus GHz. Pero seamos realistas: un Intel Core
# i7 cuesta mucho más que un Intel Celeron, aunque ambos sean Intel y tengan los mismos GHz.

# El truco: Extraer si la CPU es i3, i5, i7, Ryzen, Celeron, etc. Esto es el indicador de precio más fuerte después
# de la RAM.

# 2. La Densidad de Píxeles (PPI - Pixels Per Inch)
# Hemos sacado la resolución (X_res e Y_res). Y en tu dataset original tienes una columna llamada Inches (pulgadas)
# Si combinas ambas cosas, puedes crear una variable matemática nueva llamada PPI.

# El truco: Una pantalla Full HD de 13 pulgadas se ve mucho más nítida (y cara) que una Full HD de 17 pulgadas
# Una sola columna que calcule el PPI (√(X² + Y²) / Pulgadas) le dará a tu modelo un chivato directo de la calidad
# visual del portátil.
```
